```python
# ============================================
# Proyecto Machine Learning - Titanic Dataset
# Modelo: Logistic Regression
# División: 80% Train / 20% Test
# ============================================


# 1. Importar librerías

import pandas as pd
import numpy as np

from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression

from sklearn.metrics import (
    accuracy_score,
    confusion_matrix,
    classification_report
)


# ============================================
# 2. Cargar Dataset
# ============================================

# Descargar previamente train.csv de Kaggle Titanic
# https://www.kaggle.com/competitions/titanic

df = pd.read_csv("Titanic-Dataset.csv")


# Mostrar primeras filas

print(df.head())

print("\nInformación del dataset")
print(df.info())


# ============================================
# 3. Limpieza de datos
# ============================================

# Eliminar columnas que no aportan al modelo

df = df.drop(
    [
        "PassengerId",
        "Name",
        "Ticket",
        "Cabin"
    ],
    axis=1
)


# Revisar valores faltantes

print("\nValores faltantes")
print(df.isnull().sum())


# Completar edad faltante con la mediana

df["Age"] = df["Age"].fillna(
    df["Age"].median()
)


# Completar embarque faltante

df["Embarked"] = df["Embarked"].fillna(
    df["Embarked"].mode()[0]
)


# ============================================
# 4. Convertir variables categóricas
# ============================================


# Convertir Sex y Embarked a números

df = pd.get_dummies(
    df,
    columns=[
        "Sex",
        "Embarked"
    ],
    drop_first=True
)


print("\nDataset preparado")
print(df.head())


# ============================================
# 5. Separar variables predictoras y objetivo
# ============================================


# Variable objetivo

y = df["Survived"]


# Variables independientes

X = df.drop(
    "Survived",
    axis=1
)


# ============================================
# 6. División Train/Test
# 80% entrenamiento
# 20% prueba
# ============================================


X_train, X_test, y_train, y_test = train_test_split(

    X,
    y,

    test_size=0.20,

    random_state=42

)


print("\nDatos entrenamiento:")
print(X_train.shape)


print("\nDatos prueba:")
print(X_test.shape)



# ============================================
# 7. Crear modelo Logistic Regression
# ============================================


modelo = LogisticRegression(

    max_iter=1000

)



# Entrenar modelo

modelo.fit(
    X_train,
    y_train
)



# ============================================
# 8. Realizar predicciones
# ============================================


y_pred = modelo.predict(
    X_test
)



# ============================================
# 9. Evaluación del modelo
# ============================================


accuracy = accuracy_score(
    y_test,
    y_pred
)


print("\nAccuracy del modelo:")
print(round(accuracy,4))


print("\nMatriz de confusión:")

print(
    confusion_matrix(
        y_test,
        y_pred
    )
)


print("\nReporte de clasificación:")

print(
    classification_report(
        y_test,
        y_pred
    )
)



# ============================================
# 10. Importancia de variables
# ============================================


coeficientes = pd.DataFrame(

    {
        "Variable": X.columns,
        "Impacto": modelo.coef_[0]
    }

)


coeficientes = coeficientes.sort_values(
    by="Impacto",
    ascending=False
)


print("\nVariables más importantes:")

print(coeficientes)

# ============================================
# 11. Métricas adicionales:
# Recall y F1-score
# ============================================


from sklearn.metrics import (
    recall_score,
    f1_score
)


# Calcular Recall

recall = recall_score(
    y_test,
    y_pred
)


# Calcular F1-score

f1 = f1_score(
    y_test,
    y_pred
)


print("\nRecall del modelo:")

print(
    round(recall,4)
)


print("\nF1-score del modelo:")

print(
    round(f1,4)
)



# ============================================
# 12. Medición de latencia del modelo
# Tiempo de predicción
# ============================================


import time


# Inicio del contador

inicio = time.time()


# Ejecutar predicción

y_pred_latency = modelo.predict(
    X_test
)


# Final del contador

fin = time.time()


# Calcular latencia

latencia_total = fin - inicio


# Latencia promedio por registro

latencia_promedio = (
    latencia_total / len(X_test)
)


print("\nLatencia total de predicción:")

print(
    round(latencia_total,6),
    "segundos"
)


print("\nLatencia promedio por pasajero:")

print(
    round(latencia_promedio,8),
    "segundos"
)



# ============================================
# 13. Resumen final del modelo
# ============================================


resultados = pd.DataFrame(

    {
        "Métrica": [
            "Accuracy",
            "Recall",
            "F1-score",
            "Latencia total (seg)"
        ],

        "Resultado": [
            accuracy,
            recall,
            f1,
            latencia_total
        ]
    }

)


print("\nResumen del desempeño del modelo:")

print(resultados)
```
