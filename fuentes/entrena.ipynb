```python
# Crear estructura del proyecto

# ============================================
# Proyecto Machine Learning - Titanic Dataset
# Modelo: Random Forest
# Todo en un solo programa
# ============================================


import os


# Crear carpetas

os.makedirs(
    "datos/datos_ini",
    exist_ok=True
)


os.makedirs(
    "datos/datos_limp",
    exist_ok=True
)


print("Estructura creada")

# ============================================
# Librerías
# ============================================


import pandas as pd
import numpy as np

import time


from sklearn.model_selection import train_test_split


from sklearn.ensemble import RandomForestClassifier


from sklearn.metrics import (
    accuracy_score,
    confusion_matrix,
    classification_report,
    recall_score,
    f1_score
)


print("Librerías cargadas")

# Subir dataset Titanic

# Mover dataset original

!mv Titanic-Dataset.csv datos/datos_ini/


print("Dataset guardado")

# Cargar datos originales

# ============================================
# Cargar dataset
# ============================================


df = pd.read_csv(
    "datos/datos_ini/Titanic-Dataset.csv"
)


print(df.head())


print("\nInformación")


df.info()

# ============================================
# Limpieza
# ============================================


df_limpio = df.copy()



# Eliminar columnas innecesarias


df_limpio = df_limpio.drop(

    [
        "PassengerId",
        "Name",
        "Ticket",
        "Cabin"

    ],

    axis=1

)



# Edad faltante


df_limpio["Age"] = df_limpio["Age"].fillna(

    df_limpio["Age"].median()

)



# Embarque faltante


df_limpio["Embarked"] = df_limpio["Embarked"].fillna(

    df_limpio["Embarked"].mode()[0]

)



# Variables categóricas


df_limpio = pd.get_dummies(

    df_limpio,

    columns=[
        "Sex",
        "Embarked"
    ],

    drop_first=True

)



print(df_limpio.head())

# ============================================
# Guardar datos limpios
# ============================================


df_limpio.to_csv(

    "datos/datos_limp/Titanic-Limpio.csv",

    index=False

)


print("Dataset limpio guardado")

# ============================================
# Variables X y Y
# ============================================


y = df_limpio["Survived"]


X = df_limpio.drop(

    "Survived",

    axis=1

)


print(X.shape)

print(y.shape)

# División entrenamiento/prueba 80/20

# ============================================
# Train / Test
# ============================================


X_train, X_test, y_train, y_test = train_test_split(

    X,

    y,

    test_size=0.20,

    random_state=42

)


print(
"Datos entrenamiento:",
X_train.shape
)


print(
"Datos prueba:",
X_test.shape
)

# Crear modelo Random Forest

# ============================================
# Modelo
# ============================================


modelo = RandomForestClassifier(

    n_estimators=100,

    random_state=42

)



modelo.fit(

    X_train,

    y_train

)



print("Modelo entrenado")

# ============================================
# Predicción
# ============================================


y_pred = modelo.predict(

    X_test

)


print("Predicciones realizadas")

# ============================================
# Accuracy
# ============================================


accuracy = accuracy_score(

    y_test,

    y_pred

)



print(
"Accuracy:",
round(
accuracy,
4
)

)

# ============================================
# Confusion Matrix
# ============================================


print(

confusion_matrix(

    y_test,

    y_pred

)

)

# ============================================
# Classification Report
# ============================================


print(

classification_report(

    y_test,

    y_pred

)

)

# ============================================
# Recall y F1
# ============================================


recall = recall_score(

    y_test,

    y_pred

)


f1 = f1_score(

    y_test,

    y_pred

)



print(
"Recall:",
round(recall,4)
)



print(
"F1 Score:",
round(f1,4)
)

# ============================================
# Importancia variables
# ============================================


importancias = pd.DataFrame(

{

"Variable": X.columns,

"Importancia": modelo.feature_importances_

}

)



importancias = importancias.sort_values(

by="Importancia",

ascending=False

)



print(importancias)

# ============================================
# Latencia
# ============================================


inicio = time.time()



modelo.predict(

    X_test

)



fin = time.time()



latencia_total = fin - inicio



latencia_promedio = (

latencia_total /

len(X_test)

)



print(
"Latencia total:",
round(latencia_total,6),
"segundos"
)



print(
"Latencia promedio:",
round(latencia_promedio,8),
"segundos"
)

# ============================================
# Resumen
# ============================================


resultados = pd.DataFrame(

{

"Métrica":[

"Accuracy",

"Recall",

"F1-score",

"Latencia"

],


"Resultado":[

accuracy,

recall,

f1,

latencia_total

]

}

)


```
