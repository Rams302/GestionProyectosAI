```python
# Crear estructura del proyecto

import os


os.makedirs("datos/datos_ini", exist_ok=True)

os.makedirs("datos/datos_limp", exist_ok=True)

os.makedirs("fuentes", exist_ok=True)


print("Estructura creada correctamente")

!cp ./Titanic-Dataset.csv ./datos/datos_ini/.

import pandas as pd





def cargar_datos(ruta):

    """
    Carga dataset original
    """

    df = pd.read_csv(ruta)

    return df




def limpiar_datos(df):

    """
    Limpieza y transformación Titanic Dataset
    """

    # eliminar columnas sin valor predictivo

    df = df.drop(

        [
            "PassengerId",
            "Name",
            "Ticket",
            "Cabin"

        ],

        axis=1

    )


    # valores faltantes Age

    df["Age"] = df["Age"].fillna(

        df["Age"].median()

    )


    # valores faltantes Embarked

    df["Embarked"] = df["Embarked"].fillna(

        df["Embarked"].mode()[0]

    )



    # variables categóricas

    df = pd.get_dummies(

        df,

        columns=[

            "Sex",
            "Embarked"

        ],

        drop_first=True

    )


    return df

# =====================================
# Preparación Dataset Titanic
# =====================================



# cargar datos originales

df = cargar_datos(

"./datos/datos_ini/Titanic-Dataset.csv"

)



print(df.head())



# limpieza

df_limpio = limpiar_datos(df)



print(
df_limpio.head()
)



# guardar dataset limpio


df_limpio.to_csv(

"./datos/datos_limp/Titanic-Limpio.csv",

index=False

)



print("Datos limpios generados")
```
